```python
from yahooquery import Ticker
import pandas as pd

# ======== 1. Parameter Saham dan Periode ========
ticker = 'TLKM.JK'  # Ganti sesuai kebutuhan
start_date = '2024-12-13'
end_date = '2025-06-12'

# ======== 2. Ambil Data Harga Saham Harian ========
t = Ticker(ticker)
data = t.history(start=start_date, end=end_date, interval='1d')

# ======== 3. Reset Index (karena multi-index dari yahooquery) ========
data = data.reset_index()

# ======== 4. Filter hanya data dari ticker yang diminta (jika ambil banyak) ========
data = data[data['symbol'] == ticker]

# # ======== 5. Simpan ke CSV (Opsional) ========
# data.to_csv(f'{ticker}_periode.csv', index=False)

# ======== 6. Tampilkan 5 Baris Pertama ========
print(data.head())
```

```python
data['return'] = data['close'].pct_change().shift(1)
data['volatility'] = data['return'].rolling(window=5).std()
data['ma'] = data['close'].shift(1).rolling(window=10).mean()

data = data.dropna()
```
